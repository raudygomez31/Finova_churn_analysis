```python
import pandas as pd

# Cargar el archivo de datos
df = pd.read_csv('Nuevo Proyecto.csv')

# Ver las primeras 5 filas para asegurar que todo cargó bien
df.head()
```

```python
# Seleccionar solo las columnas numéricas que nos interesan para el negocio
columnas_clave = ['Age', 'Balance', 'NumOfProducts', 'EstimatedSalary', 'Exited']

# Calcular la matriz de correlación matemática
correlacion = df[columnas_clave].corr()

# Mostrar la matriz en pantalla
correlacion
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Configurar el tamaño del gráfico
plt.figure(figsize=(8, 6))

# Crear el mapa de calor con los datos de correlación
sns.heatmap(correlacion, annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5)

# Añadir un título profesional
plt.title('Matriz de Correlación: Factores de Abandono en Finova Pay', fontsize=14)

# Mostrar el gráfico en pantalla
plt.show()
```
